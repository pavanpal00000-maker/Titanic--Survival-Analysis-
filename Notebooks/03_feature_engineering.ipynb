# Feature Engineering

```python
# Import Libraries
import pandas as pd
import numpy as numpy

# The preprocessing module contains tools to prepare data before training a machine learning model.
from sklearn.preprocessing import LabelEncoder  # this library help to convert categorical column into numerical column
from sklearn.preprocessing import StandardScaler # It scales numerical values so that all features are on a similar scale.
```

```python
# Load Clean Dataset
df = pd.read_csv('../data/train_clean.csv')
df.head()
```

```python
df.columns
```

```python
# Drop Unnecessary Columns
df = df.drop(columns=['Ticket','Name','PassengerId'])
df.head()
```

```python
# we need encode Sex column to train machine learning model
df['Sex'].value_counts()
```

```python
# after encoding 
encoder = LabelEncoder()
df['Sex'] = encoder.fit_transform(df['Sex']) # LabelEncoder assigns labels alphabetically so Female → 0, male → 1
df.head()
```

```python
# now we do same for Embarked
df['Embarked'].unique()
```

```python
# encoding for embarked column
df['Embarked'] = encoder.fit_transform(df['Embarked']) 
df['Embarked'].head() # where C → 0, Q → 1, S → 2
```

```python
# check datatypes 
df.info()
```

```python
# Machine learning models often perform better when numerical features are on a similar scale.
numeric_columns = ['Age','Fare','SibSp','Parch']
```

```python
# Creating the scaler
scalar = StandardScaler() # It transforms values so that:
                          # Mean ≈ 0
                          # Standard deviation ≈ 1
```

```python
# The relative differences remain, but the scale is standardized.
df[numeric_columns] = scalar.fit_transform(df[numeric_columns])
```

```python
df.head()
```

```python
# Save the Processed Dataset
df.to_csv("../data/train_processed.csv", index=False)
```
