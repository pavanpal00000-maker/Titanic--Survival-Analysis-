# Bivariate Analysis

```python
# Survivied VS Gender
```

```python
# Cross table
# This shows how many males and females survived and did not survive.
pd.crosstab(df['Sex'],df['Survived'])
```

```python
# calculates the survival percentage for each gender.
pd.crosstab(df['Sex'],
         df['Survived'],
         normalize = 'index'
         ) * 100
```

```python
# ploting graph using countplot graph
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    x='Sex',
    hue = 'Survived'
)
plt.title('Survived Vs Gender')
plt.xlabel('Gender')
plt.ylabel('Number Of Passenger')
plt.legend(title = 'Survived') # it's set the legend title Survived which we can see in the right corner of the graph
plt.show()
```

# Observation

- Female passengers had a much higher survival rate than male passengers.
- Most male passengers did not survive.

# Insight

Gender was one of the strongest factors influencing survival.
Women were given priority during evacuation, resulting in a significantly higher survival rate.

# Survival vs Passenger Class

```python
pd.crosstab(df['Pclass'],df['Survived'])
```

```python
# In percentage
pd.crosstab(df['Pclass'],df['Survived'], normalize = 'index') * 100
```

```python
# ploting graph using countplot graph
plt.figure(figsize=(10,6))   # set figure 

sns.countplot(
    data = df,
    x = 'Pclass',
    hue = 'Survived'
)

plt.title('Survived by Passenger Class')
plt.xlabel('Passenger class')
plt.ylabel('Number of Passengers')
plt.show()
```

# Observation

- First-class passengers had the highest survival rate.
- Third-class passengers had the lowest survival rate.

# Insight

Passenger class had a significant impact on survival.
Higher-class passengers likely had better access to lifeboats and crew assistance.

# Survival vs Age

```python
# creating a histogram for Age Vs Survival analysis 
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x='Age',
    hue = "Survived",
    kde=True,
    bins = 30,
    multiple = 'stack'
)
plt.title('Age Distribution by survival')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
# ploting a box for analysis
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x='Survived',
    y= 'Age'
)
plt.title('Age vs Survival')
plt.show()
```

# Observation

- Survivors and non-survivors have overlapping age distributions.
- Children appear to have a relatively higher survival rate than many adults.
