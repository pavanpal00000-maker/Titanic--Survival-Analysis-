## Univeriate analysis

```python
# Checking Survival Distribution
df['Survived'].value_counts()
```

```python
# What percentage of passengers survived?
df['Survived'].value_counts(normalize=True)*100
```

### Observation
# Approximately 38.38% of passengers survived, while 61.61% did not survive.

```python
# ploting survival distribution
plt.figure(figsize=(6,4))

sns.countplot(data=df,x='Survived')
plt.title('Survival Count')
plt.xlabel('Survived')
plt.ylabel('Number Of Passenger')
plt.show()
```

### Insight

- Most passengers did not survive.
- The dataset is slightly imbalanced toward non-survivors.

```python
# Statistical Status of Age
df['Age'].describe()
```
